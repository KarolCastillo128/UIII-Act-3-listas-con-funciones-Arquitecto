#### UIII Act 3 listas con funciones agregarArquitecto() leerArquitecto(), listarArquitecto(), buscarArquitecto() , modificarArquitecto() y borrarArquitecto(),

```python
class Arquitecto:
    def __init__(self, cedula, nombre, especialidad):
        # Inicializa los atributos del arquitecto
        self.cedula = cedula
        self.nombre = nombre
        self.especialidad = especialidad

    def __str__(self):
        # Retorna una cadena con la información
        return f"Cédula: {self.cedula} | Nombre: {self.nombre} | Especialidad: {self.especialidad}"


# Lista global
lista_arquitectos = []


def leerarquitecto():
    # Registrar arquitecto
    print("\n--- REGISTRAR ARQUITECTO ---")

    cedula = input("Ingrese la cédula profesional: ")
    nombre = input("Ingrese el nombre completo: ")
    especialidad = input("Ingrese la especialidad: ")

    nuevo = Arquitecto(cedula, nombre, especialidad)
    lista_arquitectos.append(nuevo)

    print("✅ Arquitecto registrado exitosamente.")


def listararquitecto():
    # Mostrar arquitectos
    print("\n--- LISTA DE ARQUITECTOS ---")

    if not lista_arquitectos:
        print("⚠️ No hay arquitectos registrados.")
    else:
        for i, arq in enumerate(lista_arquitectos, start=1):
            print(f"{i}. {arq}")


def buscararquitecto():
    # Buscar arquitecto por cédula
    cedula = input("Ingrese la cédula a buscar: ")

    for arq in lista_arquitectos:
        if arq.cedula == cedula:
            return arq

    return None


def modificararquitecto():
    # Modificar datos
    print("\n--- MODIFICAR ARQUITECTO ---")

    arq = buscararquitecto()

    if arq:
        print(f"Arquitecto encontrado: {arq}")

        arq.nombre = input("Nuevo nombre: ")
        arq.especialidad = input("Nueva especialidad: ")

        print("✅ Datos actualizados.")
    else:
        print("❌ Arquitecto no encontrado.")


def borrararquitecto():
    # Eliminar arquitecto
    print("\n--- ELIMINAR ARQUITECTO ---")

    arq = buscararquitecto()

    if arq:
        lista_arquitectos.remove(arq)
        print("🗑️ Registro eliminado correctamente.")
    else:
        print("❌ No se encontró el registro.")


# Programa principal
while True:

    print("\n========== SISTEMA DE GESTIÓN ==========")
    print("1. Registrar arquitecto")
    print("2. Listar arquitectos")
    print("3. Buscar arquitecto")
    print("4. Modificar arquitecto")
    print("5. Eliminar arquitecto")
    print("6. Salir")
    print("========================================")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        leerarquitecto()

    elif opcion == "2":
        listararquitecto()

    elif opcion == "3":
        print("\n--- BUSCAR ARQUITECTO ---")

        resultado = buscararquitecto()

        if resultado:
            print("✅ Arquitecto encontrado:")
            print(resultado)
        else:
            print("❌ Arquitecto no encontrado.")

    elif opcion == "4":
        modificararquitecto()

    elif opcion == "5":
        borrararquitecto()

    elif opcion == "6":
        print("👋 Saliendo del sistema...")
        break

    else:
        print("⚠️ Opción inválida, intente de nuevo.")
```
